```python
## handling the missing values, replacing the NaN with 0

cols = ['total_ic_mou_9','total_og_mou_9','vol_2g_mb_9','vol_3g_mb_9']

hvc_df[cols] = hvc_df[cols].fillna(0)
```

```python
hvc_df[cols]
```

```python
## churn column creation
hvc_df['churn'] = (
    (hvc_df['total_ic_mou_9'] == 0) &
    (hvc_df['total_og_mou_9'] == 0) &
    (hvc_df['vol_2g_mb_9'] == 0) &
    (hvc_df['vol_3g_mb_9'] == 0)
).astype(int)
```

```python
## checking the churn distribution

hvc_df['churn'].value_counts()
```

```python
## checking the %

hvc_df['churn'].value_counts(normalize=True) * 100
```

```python
## removing all the month 9 columns, to prevent data leakage

month9_cols = [col for col in hvc_df.columns if '_9' in col]   ##finding all the month 9 columns

hvc_df = hvc_df.drop(columns=month9_cols) ## deleting them
```

```python
##verifying the removal

print(len(month9_cols), "columns removed")
print(hvc_df.shape)
```
