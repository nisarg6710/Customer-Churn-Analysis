```python
import pandas as pd
```

```python
## loading the phase 2 dataset
df = pd.read_csv('../data/phase2_churn_label_creation.csv')
```

#### Decrease in usage

```python
## outgoing usage drop
df['og_usage_avg_6_7'] = (df['total_og_mou_6'] + df['total_og_mou_7']) / 2

df['og_usage_drop'] = df['total_og_mou_8'] - df['og_usage_avg_6_7']


## incoming usage drop
df['ic_usage_avg_6_7'] = (df['total_ic_mou_6'] + df['total_ic_mou_7']) / 2

df['ic_usage_drop'] = df['total_ic_mou_8'] - df['ic_usage_avg_6_7']
```

```python
## to find the data usage now we will be combining the usage of 2g and 3g across months 6,7 and 8

df['data_usage_6'] = df['vol_2g_mb_6'] + df['vol_3g_mb_6']
df['data_usage_7'] = df['vol_2g_mb_7'] + df['vol_3g_mb_7']
df['data_usage_8'] = df['vol_2g_mb_8'] + df['vol_3g_mb_8']
```

```python
##avg of month 6 and 7

df['data_avg_6_7'] = (df['data_usage_6'] + df['data_usage_7']) / 2
```

```python
## now check for usage change

df['data_usage_drop'] = df['data_usage_8'] - df['data_avg_6_7']
```

#### Feature 2 — Recharge Frequency Drop

```python
## Customers who churn often stop recharging frequently

## avg recharge count

df['rech_cnt_avg_6_7'] = (df['total_rech_num_6'] + df['total_rech_num_7']) / 2
```

```python
## -ve value means customer is recharging less often.

df['rech_freq_drop'] = df['total_rech_num_8'] - df['rech_cnt_avg_6_7']
```

```python
df.head()
```

```python
df.tail()
```

#### Feature 3 — Recharge Amount Drop

```python
## Customers may start spending less.

df['rech_amt_avg_6_7'] = (df['total_rech_amt_6'] + df['total_rech_amt_7']) / 2

df['rech_amt_drop'] = df['total_rech_amt_8'] - df['rech_amt_avg_6_7']
```

```python
df.head()
```

#### Feature 4 — Data vs Voice Behaviour

```python
## maybe data usage drops but voice stays stable.

## compute voice average

df['voice_usage_6'] = df['total_og_mou_6'] + df['total_ic_mou_6']
df['voice_usage_7'] = df['total_og_mou_7'] + df['total_ic_mou_7']
df['voice_usage_8'] = df['total_og_mou_8'] + df['total_ic_mou_8']

df['voice_avg_6_7'] = (df['voice_usage_6'] + df['voice_usage_7']) / 2
```

```python
## voice change

df['voice_usage_drop'] = df['voice_usage_8'] - df['voice_avg_6_7']
```

```python
## data v/s voice ratio
## +1 is the correction factor, helps in handling 0 in Denominator.

df['data_voice_ratio_8'] = df['data_usage_8'] / (df['voice_usage_8'] + 1)
```

```python
## handling the missing values

df = df.fillna(0)
```
